# Exploratory Data Analysis — Adult Census Income

Quick look at the dataset: distributions, missing values, and relationships with the target.

```python
import sys
sys.path.append('..')

import matplotlib.pyplot as plt
import seaborn as sns

from src.data_loader import load_train_data

df = load_train_data()
df.head()
```

```python
df.info()
df.describe()
```

```python
df['income'].value_counts(normalize=True).plot(kind='bar', title='Target class balance')
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 5))
sns.histplot(data=df, x='age', hue='income', multiple='stack', ax=ax)
ax.set_title('Age distribution by income class')
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 5))
sns.countplot(data=df, y='education', hue='income', order=df['education'].value_counts().index, ax=ax)
ax.set_title('Education level vs income class')
plt.show()
```
